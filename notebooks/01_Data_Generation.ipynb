# 1. Import Libraries
In this section, we import the necessary libraries. We use `pandas` for data manipulation, `numpy` and `random` for generating random data, and `Faker` to generate realistic names and dates.

```python
# ==============================================
# Student Learning Analytics
# Synthetic Dataset Generation
# ==============================================

import pandas as pd
import numpy as np
import random
import os
from faker import Faker

fake = Faker("en_IN")

random.seed(42)
np.random.seed(42)
```

# 2. Create Data Folder
Here we create a `data` directory in the parent folder if it doesn't already exist. This is where all the generated CSV files will be stored.

```python
# Create data folder if it doesn't exist

os.makedirs("../data", exist_ok=True)
```

# 3. Number of Students
We define the total number of students for whom we will generate the synthetic dataset.

```python
NUM_STUDENTS = 2000
```

# 4. Define Possible Values
We define the possible categories for various attributes such as Classes, Boards, Target Exams, Mediums of Instruction, and Study Modes.

```python
classes = [
    "Class 9",
    "Class 10",
    "Class 11",
    "Class 12"
]

boards = [
    "CBSE",
    "ICSE",
    "State Board"
]

target_exams = [
    "JEE",
    "NEET",
    "Foundation"
]

mediums = [
    "English",
    "Tamil",
    "Hindi"
]

study_modes = [
    "Online",
    "Offline",
    "Hybrid"
]
```

# 5. Generate Students Dataset
We generate the core student records. For each student, we assign random categorical values and generate a unique Student ID and Name using the Faker library.

```python
students = []

for i in range(1, NUM_STUDENTS + 1):

    student = {

        "Student_ID": f"STU{i:04d}",

        "Student_Name": fake.name(),

        "Class": random.choice(classes),

        "Board": random.choice(boards),

        "Target_Exam": random.choice(target_exams),

        "Medium": random.choice(mediums),

        "Admission_Date": fake.date_between(
            start_date="-2y",
            end_date="today"
        ),

        "Study_Mode": random.choice(study_modes)

    }

    students.append(student)
```

# 6. Save Students Dataset
We convert the list of student dictionaries into a Pandas DataFrame, inspect it, and save it to `data/students.csv`.

```python
students_df = pd.DataFrame(students)
```

```python
students_df.to_csv(
    "../data/students.csv",
    index=False
)
```

```python
student_ids = students_df["Student_ID"].tolist()
```

```python
subjects = [
    "Physics",
    "Chemistry",
    "Mathematics",
    "Biology"
]

chapters = {
    "Physics": [
        "Motion",
        "Laws of Motion",
        "Work and Energy",
        "Electricity"
    ],
    "Chemistry": [
        "Atoms",
        "Chemical Bonding",
        "Acids and Bases",
        "Organic Chemistry"
    ],
    "Mathematics": [
        "Algebra",
        "Trigonometry",
        "Calculus",
        "Probability"
    ],
    "Biology": [
        "Cell",
        "Genetics",
        "Human Body",
        "Plant Physiology"
    ]
}

faculty = [
    "Dr. Kumar",
    "Dr. Priya",
    "Dr. Arun",
    "Dr. Meena",
    "Dr. Ravi"
]
```

## 7. Generate Attendance Records
We generate attendance data. For each student, we create 10 random lecture attendance records across different subjects and chapters.

```python
attendance = []

lecture_id = 1

for student in student_ids:

    for _ in range(10):

        subject = random.choice(subjects)

        attendance.append({

            "Student_ID": student,

            "Lecture_ID": f"LEC{lecture_id:04d}",

            "Subject": subject,

            "Chapter": random.choice(chapters[subject]),

            "Faculty": random.choice(faculty),

            "Lecture_Date": fake.date_between(
                start_date="-6M",
                end_date="today"
            ),

            "Attendance_Percentage": random.randint(50,100)

        })

        lecture_id += 1
```

```python
attendance_df = pd.DataFrame(attendance)

attendance_df.head()
```

```python
attendance_df.to_csv(
    "../data/attendance.csv",
    index=False
)

print("attendance.csv created successfully!")
```

## 8. Generate Video Logs
We simulate video watching behavior. We generate 8 video logs per student, calculating watch times and completion statuses based on the total video length.

```python
video_logs = []

video_id = 1

for student in student_ids:

    for _ in range(8):

        subject = random.choice(subjects)

        video_length = random.randint(20,90)

        watch_time = random.randint(10,video_length)

        completion = round((watch_time/video_length)*100,2)

        if completion >= 90:
            status = "Completed"
        elif completion >= 50:
            status = "Partially Completed"
        else:
            status = "Not Completed"

        video_logs.append({

            "Student_ID": student,

            "Video_ID": f"VID{video_id:04d}",

            "Subject": subject,

            "Chapter": random.choice(chapters[subject]),

            "Video_Length": video_length,

            "Watch_Time": watch_time,

            "Percentage_Completed": completion,

            "Number_of_Replays": random.randint(0,5),

            "Completion_Status": status

        })

        video_id += 1
```

```python
video_df = pd.DataFrame(video_logs)

video_df.head()
```

```python
video_df.to_csv(
    "../data/video_logs.csv",
    index=False
)

print("video_logs.csv created successfully!")
```

## 9. Generate Quiz Attempts
We simulate quiz performances. For each student, we generate 6 quiz attempts, randomizing the number of questions, correct answers, and calculating the accuracy.

```python
quiz_attempts = []

quiz_id = 1

for student in student_ids:

    for _ in range(6):

        subject = random.choice(subjects)

        questions = random.randint(15, 30)

        correct = random.randint(5, questions)

        incorrect = questions - correct

        accuracy = round((correct / questions) * 100, 2)

        quiz_attempts.append({

            "Student_ID": student,

            "Quiz_ID": f"QUIZ{quiz_id:04d}",

            "Subject": subject,

            "Chapter": random.choice(chapters[subject]),

            "Questions_Attempted": questions,

            "Correct": correct,

            "Incorrect": incorrect,

            "Accuracy": accuracy,

            "Average_Time_Per_Question": round(random.uniform(20, 90), 2)

        })

        quiz_id += 1
```

```python
quiz_df = pd.DataFrame(quiz_attempts)

quiz_df.head()
```

```python
quiz_df.to_csv("../data/quiz_attempts.csv", index=False)

print("quiz_attempts.csv created successfully!")
```

## 10. Generate Mock Tests
We create mock test records. Each student takes 3 mock tests, and we simulate marks, ranks, negative marks, and time utilization.

```python
mock_tests = []

test_id = 1

for student in student_ids:

    for _ in range(3):

        subject = random.choice(subjects)

        marks = random.randint(20, 100)

        rank = random.randint(1, 2000)

        negative = round(random.uniform(0, 15), 2)

        skipped = random.randint(0, 10)

        utilization = random.randint(60, 100)

        mock_tests.append({

            "Student_ID": student,

            "Test_ID": f"TEST{test_id:04d}",

            "Subject": subject,

            "Marks": marks,

            "Rank": rank,

            "Negative_Marks": negative,

            "Questions_Skipped": skipped,

            "Time_Utilization": utilization

        })

        test_id += 1
```

```python
mock_df = pd.DataFrame(mock_tests)

mock_df.head()
```

```python
mock_df.to_csv("../data/mock_tests.csv", index=False)

print("mock_tests.csv created successfully!")
```

```python
print("Quiz Records :", len(quiz_df))
print("Mock Test Records :", len(mock_df))
```

## 11. Generate Assignments
We simulate assignment submissions. Each student has 5 assignments, and we randomize whether they were submitted, the submission delay, and the score.

```python
assignments = []

assignment_id = 1

for student in student_ids:

    for _ in range(5):

        subject = random.choice(subjects)

        submitted = random.choice(["Yes", "No"])

        if submitted == "Yes":
            delay = random.randint(0, 5)
            score = random.randint(40, 100)
        else:
            delay = random.randint(6, 15)
            score = random.randint(0, 40)

        assignments.append({

            "Student_ID": student,

            "Assignment_ID": f"ASSIGN{assignment_id:04d}",

            "Subject": subject,

            "Submitted": submitted,

            "Submission_Delay": delay,

            "Score": score

        })

        assignment_id += 1
```

```python
assignment_df = pd.DataFrame(assignments)

assignment_df.head()
```

```python
assignment_df.to_csv(
    "../data/assignments.csv",
    index=False
)

print("assignments.csv created successfully!")
```

## 12. Generate Engagement Metrics
We generate overall engagement metrics such as login frequency, weekly study hours, and average session duration for each student.

```python
engagement = []

for student in student_ids:

    engagement.append({

        "Student_ID": student,

        "Login_Frequency": random.randint(5, 30),

        "Weekly_Study_Hours": round(random.uniform(5, 35), 1),

        "Average_Session_Duration": random.randint(20, 180),

        "Consecutive_Study_Days": random.randint(1, 30),

        "Sessions_Per_Week": random.randint(2, 14)

    })
```

```python
engagement_df = pd.DataFrame(engagement)

engagement_df.head()
```

```python
engagement_df.to_csv(
    "../data/engagement.csv",
    index=False
)

print("engagement.csv created successfully!")
```
