```python
# =====================================================
# Import Required Libraries
# =====================================================

import pandas as pd
import matplotlib.pyplot as plt
```

## Load Processed Dataset

```python
df = pd.read_csv("../data/student_performance.csv")

df.head()

# Replace student_performance.csv with your processed dataset name if different.
```

# Dataset Summary

```python
print("Total Students :", len(df))

print("Total Features :", len(df.columns))

print("\nDataset Shape")

print(df.shape)
```

# Attendance Analysis

```python
plt.figure(figsize=(8,5))

df["Attendance_Percentage"].hist(bins=10)

plt.title("Attendance Distribution")

plt.xlabel("Attendance (%)")

plt.ylabel("Number of Students")

plt.show()
```

### Insight

Students with higher attendance generally perform better in assessments.

Low attendance may indicate students who need additional academic support.

# Assignment Performance

```python
plt.figure(figsize=(8,5))

df["Assignment_Average"].hist(bins=10)

plt.title("Assignment Performance")

plt.xlabel("Assignment Score")

plt.ylabel("Students")

plt.show()
```

### Insight

Students who consistently complete assignments tend to achieve better overall performance.

# Quiz Performance

```python
plt.figure(figsize=(8,5))

df["Quiz_Average"].hist(bins=10)

plt.title("Quiz Performance")

plt.xlabel("Quiz Score")

plt.ylabel("Students")

plt.show()
```

### Insight

Regular quiz participation helps identify learning gaps early and improves academic progress.

# Engagement Analysis

```python
plt.figure(figsize=(8,5))

df["Engagement_Score"].hist(bins=10)

plt.title("Student Engagement")

plt.xlabel("Engagement Score")

plt.ylabel("Students")

plt.show()
```

### Insight

Higher engagement scores are associated with better academic performance and participation.

# Students At Risk

```python
at_risk = df[df["Attendance_Percentage"] < 75]

print("Students At Risk :", len(at_risk))

at_risk.head()

# Adjust the threshold or column name if your dataset uses a different definition.
```

### Recommendation

Students with attendance below 75% should receive:

- Academic counselling
- Attendance monitoring
- Personalized learning support

# Overall Findings

```python
summary = {
    "Total Students": len(df),
    "Average Attendance": round(df["Attendance_Percentage"].mean(), 2),
    "Average Assignment Score": round(df["Assignment_Average"].mean(), 2),
    "Average Quiz Score": round(df["Quiz_Average"].mean(), 2),
    "Average Engagement": round(df["Engagement_Score"].mean(), 2)
}

summary
```

# Recommendations

Based on the analysis, the following actions are recommended:

1. Monitor students with low attendance.
2. Encourage regular assignment submissions.
3. Promote active participation in quizzes.
4. Increase student engagement through interactive learning activities.
5. Use predictive analytics to identify at-risk students early.
